```python
import pandas as pd
import numpy as np

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 10)
```

- Definimos las rutas

```python
from pathlib import Path

BASE = Path("../data")

RUTA_REE = BASE / "ree"
RUTA_AEMET = BASE / "aemet"
RUTA_ESIOS = BASE / "esios"
RUTA_DATADIS = BASE / "datadis"
RUTA_PVGIS = BASE / "pvgis"
```

- Leemos todos los archivos

```python
demanda = pd.read_csv(RUTA_REE / "demanda_horaria.csv")

generacion = pd.read_csv(RUTA_REE / "generacion_estructura.csv")

renovables = pd.read_csv(RUTA_REE / "renovable_vs_no_renovable.csv")

clima = pd.read_csv(RUTA_AEMET / "climatologia_diaria.csv")

solar = pd.read_csv(RUTA_ESIOS / "generacion_solar_fv.csv")

eolica = pd.read_csv(RUTA_ESIOS / "generacion_eolica.csv")

consumo = pd.read_csv(RUTA_DATADIS / "consumo_8639HR.csv")

radiacion = pd.read_csv(RUTA_PVGIS / "radiacion_horaria.csv")

produccion = pd.read_csv(RUTA_PVGIS / "produccion_fv_mensual.csv")
```

- Compruebamos que todo se ha cargado correctamente

```python
datasets = {
    "Demanda REE": demanda,
    "Generación REE": generacion,
    "Renovables REE": renovables,
    "Clima AEMET": clima,
    "Solar ESIOS": solar,
    "Eólica ESIOS": eolica,
    "Consumo Datadis": consumo,
    "Radiación PVGIS": radiacion,
    "Producción PVGIS": produccion
}

for nombre, df in datasets.items():
    print(f"\n===== {nombre} =====")
    print(df.shape)
    print(df.head(3))
```

```python
# =====================================================================
# 1. PARSEO SEGURO DE FECHAS EN TODOS LOS DATASETS (Excluyendo Mensual)
# =====================================================================
for df_name, df in datasets.items():
    if "produc" in df_name.lower(): 
        continue
    
    # Encontrar dinámicamente cualquier columna con indicios de tiempo
    col_fecha_list = [c for c in df.columns if any(x in c.lower() for x in ['fech', 'date', 'time', 'timestamp'])]
    if not col_fecha_list:
        print(f"⚠️ Alerta: No se detectó columna temporal en {df_name}")
        continue
        
    col_fecha = col_fecha_list[0]
    # Conversión a datetime eliminando zonas horarias locales para evitar conflictos de cruce
    df['datetime_clean'] = pd.to_datetime(df[col_fecha], utc=True, errors='coerce').dt.tz_localize(None)


# =====================================================================
# 2. FUNCIÓN DE EXTRACCIÓN FLEXIBLE DE LA COLUMNA DE MÉTRICA/VALOR
# =====================================================================
def extraer_metrica_segura(df_name, nuevo_nombre_col, filtrar_serie=None):
    df = datasets[df_name].copy()
    
    if 'datetime_clean' not in df.columns:
        return None
        
    # Aplicar filtros específicos si es necesario (ej: Demanda REE 'Real')
    if filtrar_serie and 'serie' in df.columns:
        df = df[df['serie'] == filtrar_serie]
        
    # Buscar dinámicamente qué columna contiene los datos numéricos
    columnas_candidatas = [c for c in df.columns if c not in ['categoria', 'serie', 'tipo', 'fecha', 'datetime_clean', 'geo_name', 'indicativo', 'nombre', 'provincia']]
    
    # Priorizar la columna 'valor' si existe, de lo contrario tomar la primera numérica válida
    if 'valor' in df.columns:
        col_metrica = 'valor'
    elif columnas_candidatas:
        col_metrica = columnas_candidatas[0]
    else:
        print(f"❌ No se encontró ninguna columna numérica para el dataset: {df_name}")
        return None
    
    # Forzamos la conversión a numérico de la métrica por si viene como texto
    df[col_metrica] = pd.to_numeric(df[col_metrica], errors='coerce')
        
    # SOLUCIÓN AL ERROR: Usamos numeric_only=True al agrupar para ignorar columnas de texto residuales
    df_resultado = df.groupby('datetime_clean')[col_metrica].mean(numeric_only=True).reset_index()
    return df_resultado.rename(columns={col_metrica: nuevo_nombre_col})


# =====================================================================
# 3. PROCESAMIENTO E INTERPOLACIÓN HORARIA INDIVIDUAL
# =====================================================================
df_demanda    = extraer_metrica_segura("Demanda REE", 'demanda_ree_kwh', filtrar_serie='Real')
df_generacion = extraer_metrica_segura("Generación REE", 'generacion_total_ree_kwh')
df_renovables = extraer_metrica_segura("Renovables REE", 'renovables_ree_kwh')
df_solar      = extraer_metrica_segura("Solar ESIOS", 'solar_esios_kwh')
df_eolica     = extraer_metrica_segura("Eólica ESIOS", 'eolica_esios_kwh')
df_consumo    = extraer_metrica_segura("Consumo Datadis", 'consumo_casa_datadis_kwh')

# Caso Especial: Radiación PVGIS (Suele almacenar múltiples parámetros como G(i) y T2m)
key_radiacion = [k for k in datasets.keys() if "radiac" in k.lower() or "radiación" in k.lower()][0]
df_radiacion_raw = datasets[key_radiacion].copy()

cols_pvgis_dict = {}
if 'G(i)' in df_radiacion_raw.columns: cols_pvgis_dict['G(i)'] = 'radiacion_solar_pvgis'
if 'T2m' in df_radiacion_raw.columns: cols_pvgis_dict['T2m'] = 'temperatura_pvgis'

if not cols_pvgis_dict:
    num_cols = [c for c in df_radiacion_raw.columns if c not in ['time', 'datetime_clean']]
    for idx, c in enumerate(num_cols[:2]):
        cols_pvgis_dict[c] = f'pvgis_meteo_var_{idx}'

# Forzamos conversión a numérico en las columnas de PVGIS
for col in cols_pvgis_dict.keys():
    df_radiacion_raw[col] = pd.to_numeric(df_radiacion_raw[col], errors='coerce')

df_radiacion = df_radiacion_raw.groupby('datetime_clean')[list(cols_pvgis_dict.keys())].mean(numeric_only=True).reset_index()
df_radiacion = df_radiacion.rename(columns=cols_pvgis_dict)


# =====================================================================
# 4. FUSIÓN MAESTRA CRONOLÓGICA (Outer Join Completo)
# =====================================================================
master_df = df_demanda

listado_horarios = [df_generacion, df_renovables, df_solar, df_eolica, df_consumo, df_radiacion]
for df_h in listado_horarios:
    if df_h is not None:
        master_df = pd.merge(master_df, df_h, on='datetime_clean', how='outer')


# =====================================================================
# 5. INTEGRACIÓN DE DATOS METEOROLÓGICOS DIARIOS (AEMET)
# =====================================================================
df_aemet = datasets["Clima AEMET"].copy()
df_aemet['fecha_diaria'] = df_aemet['datetime_clean'].dt.date
master_df['fecha_diaria'] = master_df['datetime_clean'].dt.date

columnas_clima = ['fecha_diaria', 'tmed', 'prec', 'tmin', 'tmax', 'sol', 'hrMedia']
cols_aemet_existentes = [c for c in columnas_clima if c in df_aemet.columns]

# Forzamos conversión a numérico en las columnas seleccionadas de AEMET
for col in cols_aemet_existentes:
    if col != 'fecha_diaria':
        df_aemet[col] = pd.to_numeric(df_aemet[col], errors='coerce')

df_aemet_limpio = df_aemet.groupby('fecha_diaria')[cols_aemet_existentes].mean(numeric_only=True).reset_index()
master_df = pd.merge(master_df, df_aemet_limpio, on='fecha_diaria', how='left')


# =====================================================================
# 6. INTEGRACIÓN DE DATOS MACRO MENSUALES (PVGIS Rendimiento)
# =====================================================================
key_produccion = [k for k in datasets.keys() if "produc" in k.lower()][0]
df_pvgis_m = datasets[key_produccion].copy()

col_mes = [c for c in df_pvgis_m.columns if c.lower() in ['month', 'mes']][0]
df_pvgis_m = df_pvgis_m.rename(columns={col_mes: 'month_puente'})

master_df['month_puente'] = master_df['datetime_clean'].dt.month

rename_pvgis = {}
if 'E_d' in df_pvgis_m.columns: rename_pvgis['E_d'] = 'pvgis_prod_media_diaria_kwh'
if 'E_m' in df_pvgis_m.columns: rename_pvgis['E_m'] = 'pvgis_prod_media_mensual_kwh'
df_pvgis_m = df_pvgis_m.rename(columns=rename_pvgis)

# Forzamos conversión a numérico en las columnas mensuales
for col in rename_pvgis.values():
    if col in df_pvgis_m.columns:
        df_pvgis_m[col] = pd.to_numeric(df_pvgis_m[col], errors='coerce')

cols_finales_pvgis = ['month_puente'] + [v for v in rename_pvgis.values() if v in df_pvgis_m.columns]
master_df = pd.merge(master_df, df_pvgis_m[cols_finales_pvgis], on='month_puente', how='left')


# =====================================================================
# 7. SANEAMIENTO, ORDENACIÓN Y EXPORTACIÓN DEL DATASET PROCESADO
# =====================================================================
master_df = master_df.drop(columns=['fecha_diaria', 'month_puente'])
master_df = master_df.sort_values(by='datetime_clean').reset_index(drop=True)

# Corregir vacíos breves mediante interpolación lineal de series temporales
master_df = master_df.interpolate(method='linear', limit=3)

# Almacenamiento físico en data/processed
ruta_salida = BASE / "processed" / "energia_limpia.csv"
ruta_salida.parent.mkdir(parents=True, exist_ok=True)
master_df.to_csv(ruta_salida, index=False)

print("¡ETL finalizado con éxito rotundo!")
print(f"Dimensiones finales guardadas: {master_df.shape[0]} filas unificadas.")
print("\nVariables finales que se han consolidado en 'energia_limpia.csv':")
print(master_df.columns.tolist())
```
