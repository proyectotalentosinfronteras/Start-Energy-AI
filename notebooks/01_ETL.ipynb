```python
import pandas as pd
import numpy as np

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 10)
```

- Definimos las rutas

```python
from pathlib import Path

BASE = Path("../data")

RUTA_REE = BASE / "ree"
RUTA_AEMET = BASE / "aemet"
RUTA_ESIOS = BASE / "esios"
RUTA_DATADIS = BASE / "datadis"
RUTA_PVGIS = BASE / "pvgis"
```

- Leemos todos los archivos

```python
demanda = pd.read_csv(RUTA_REE / "demanda_horaria.csv")

generacion = pd.read_csv(RUTA_REE / "generacion_estructura.csv")

renovables = pd.read_csv(RUTA_REE / "renovable_vs_no_renovable.csv")

clima = pd.read_csv(RUTA_AEMET / "climatologia_diaria.csv")

solar = pd.read_csv(RUTA_ESIOS / "generacion_solar_fv.csv")

eolica = pd.read_csv(RUTA_ESIOS / "generacion_eolica.csv")

consumo = pd.read_csv(RUTA_DATADIS / "consumo_8639HR.csv")

radiacion = pd.read_csv(RUTA_PVGIS / "radiacion_horaria.csv")

produccion = pd.read_csv(RUTA_PVGIS / "produccion_fv_mensual.csv")
```

- Compruebamos que todo se ha cargado correctamente

```python
datasets = {
    "Demanda REE": demanda,
    "Generación REE": generacion,
    "Renovables REE": renovables,
    "Clima AEMET": clima,
    "Solar ESIOS": solar,
    "Eólica ESIOS": eolica,
    "Consumo Datadis": consumo,
    "Radiación PVGIS": radiacion,
    "Producción PVGIS": produccion
}

for nombre, df in datasets.items():
    print(f"\n===== {nombre} =====")
    print(df.shape)
    print(df.head(3))
```
